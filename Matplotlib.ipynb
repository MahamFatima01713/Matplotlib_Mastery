```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#There are two types of data 
#Numerical data and Catagorical data.
#Numerical data include age,marks,socres and this all type of data while Catagorical data include 
#catagories such as gender,teams,institute name etc
#FRIST GRAPH:
#First grap is 2D line plot.This graph is use for bavariate analysis(if we want to analyze two
#columns simultaneously).u Cant plot 2d plot on a single column.
#U can apply on the following scenerios:
#when both columns are numerical or both are catagorical
#when one column is numericak or other is catagorical
#In general we apply on time series data 
price=[20000,30000,40000,50000,550000]
year=[2015,2016,2017,2018,2019]
plt.plot(year,price)
```

```python
batsman=pd.read_csv('sharma-kohli.csv')
batsman
```

```python
plt.plot(batsman['index'],batsman['V Kohli'])
```

```python
#Plotting multiple plots on diffrent columns of same graph:
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
#Blue one is virat kohli and green one is Rohit sharma
```

```python
#by following u can tell that which quantity is on X-Axis and which one is on Y-Axis.And what is 
#your graph showing
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
plt.title('Rohit Sharma VS Virat Kohli Career Comparison') #gives title to graph
plt.xlabel('season')#Gives your x-axis a name
plt.ylabel('Runs Scored')#Gives your y-axis a name
```

```python
#If u want to give the colour of your choice
# If u want to give the colour of your choice
plt.plot(batsman['index'], batsman['V Kohli'], color='green')
plt.plot(batsman['index'], batsman['RG Sharma'], color='black')
plt.title('Rohit Sharma VS Virat Kohli Career Comparison')
plt.xlabel('season')
plt.ylabel('Runs Scored')
```

```python
#Line style is use to differ the appearence of line
plt.plot(batsman['index'], batsman['V Kohli'], color='green',linestyle='dashed')
plt.plot(batsman['index'], batsman['RG Sharma'], color='black',linestyle='dotted')
plt.title('Rohit Sharma VS Virat Kohli Career Comparison')
plt.xlabel('season')
plt.ylabel('Runs Scored')
#Another one is dashdot=that gives u dash and dotted lines combine in one
```

```python
#We can also control the width of the lines by 
plt.plot(batsman['index'], batsman['V Kohli'], color='green',linestyle='dashed',linewidth='3')
plt.plot(batsman['index'], batsman['RG Sharma'], color='black',linestyle='dotted')
```

```python
#MARKERS=Points that shows highest or lowest values of an axis.So u can handle the shape of your markers
# It can be diamond(D), + sign, a "o", a > or < sign
plt.plot(batsman['index'], batsman['V Kohli'], color='green',linestyle='dashed',linewidth='3',marker='+',markersize='10')
plt.plot(batsman['index'], batsman['RG Sharma'], color='black',linestyle='dotted')
#Marker size=>u can change the size of a marker also
```

```python
#By seeing any line on graph u cant explain whcih one is showing whome.So u can mention them on graph
#by "lable".Lable will mention them.But to run lable properly u need to run a function known as legend
plt.plot(batsman['index'], batsman['V Kohli'], color='green',linestyle='dashed',linewidth='3',marker='+',markersize='5',label='V Kohli')
plt.plot(batsman['index'], batsman['RG Sharma'], color='black',linestyle='dotted',linewidth='2',marker='D',markersize='5',label=' RG Sharma')
plt.legend()
```

```python
#Why it appears on this side only??? Cuz there is a parameter which settles this where the space is
#left on your graph.This is =>loc which is by default "best".
#(loc="best").If u want it to appear on your set side u can write it as  for example (loc="right")
plt.plot(batsman['index'], batsman['V Kohli'], color='green',linestyle='dashed',linewidth='3',marker='+',markersize='5',label='V Kohli')
plt.plot(batsman['index'], batsman['RG Sharma'], color='black',linestyle='dotted',linewidth='2',marker='D',markersize='5',label=' RG Sharma')
plt.legend(loc='right')
#But it looks messy so there is not that much benefit of it
```

```python
#Now for example there is sudden change in the values of ypur graph and one among them varies highest 
#or lowest as compared to other.So in this case the variation visualization of the remaining values
#does'nt describe that much.so u can put this following and can give it a range to show u proper variation
price=[20000,30000,40000,50000,70000]
year=[2015,2016,2017,2018,2019]
plt.plot(year,price)
plt.ylim(0,80000)
```

```python
#As u see the limit on y axis u can also see for x axis
price=[20000,30000,40000,50000,70000]
year=[2015,2016,2017,2018,2019]
plt.plot(year,price)
plt.ylim(0,80000)
plt.xlim(2018,2019) 
```

```python
#Bar chart:
#It is used for bivariet analysis,when u do analysis on numerical vs catagorical columns
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
children=[10,20,30,40,50]
colour=['red','blue','yellow','white','black']
plt.bar(colour,children)
#Your can also apply Xlable,Ylable and all other funtions u plot onother graphs
```

```python
#barh=is used for horizontal bar graph
plt.barh(colour,children)
#If there are more number or catagories then their is preferable that u use horizontal bar graph
```

```python
#When u have to plot multiple catagories at a same time
df=pd.read_csv('batsman_season_record.csv')
df
```

```python
plt.bar(np.arange(df.shape[0]) - 0.2,df['2015'],width=0.2)
plt.bar(np.arange(df.shape[0]),df['2016'],width=0.2)
plt.bar(np.arange(df.shape[0]) + 0.2,df['2017'],width=0.2)
```

```python
#We use xtickes for labling,first u tell ur current label then u give with which u want to replace
plt.bar(np.arange(df.shape[0]) - 0.2,df['2015'],width=0.2)
plt.bar(np.arange(df.shape[0]),df['2016'],width=0.2)
plt.bar(np.arange(df.shape[0]) + 0.2,df['2017'],width=0.2)
plt.xticks(np.arange(df.shape[0]),df['batsman'])#1:21
```
